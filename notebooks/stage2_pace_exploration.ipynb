# Stage 2.4–2.7 — Pace Modeling & Controlled Experiments (E0–E8)
Compares point-in-time baselines (`E0`–`E3`) against the regularized additive pace models (`E4`–`E8`) across chronological validation and held-out test races.

```python
import json
from pathlib import Path
import pandas as pd

metrics = json.loads(Path('../artifacts/runs/stage2_pace_v1/metrics.json').read_text())
rows = []
for exp_id, res in metrics['experiments'].items():
    v = res.get('val_h1', {})
    t = res.get('test_h1', {})
    rows.append({
        'experiment': exp_id,
        'val_race_mae_s': v.get('race_weighted_mae_s'),
        'val_rmse_s': v.get('rmse_s'),
        'test_race_mae_s': t.get('race_weighted_mae_s'),
        'test_rmse_s': t.get('rmse_s'),
        'test_cov_50': t.get('coverage_50'),
        'test_cov_90': t.get('coverage_90'),
    })
pd.DataFrame(rows)
```
