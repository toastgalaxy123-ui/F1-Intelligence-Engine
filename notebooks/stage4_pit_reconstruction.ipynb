# Stage 4: reconstruction and reference disagreement
Run the Stage 4 CLI first. These notebooks only inspect saved artifacts.


```python
from pathlib import Path
import json
import pandas as pd
root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent
run = root / 'artifacts/runs/stage4_pit_loss_v1'
if not (run / 'metrics.json').exists():
    raise FileNotFoundError('Run f1ri stage4 from the repository root first')
```

```python
labels = pd.read_parquet(run / 'pit_event_labels.parquet')
primary = labels.query("measurement_method == 'TWO_LAP_CROSS_FIT'")
print(primary[['pit_event_id','normal_in_reference_s','normal_out_reference_s','transition_loss_s','label_std_s','matched_control_loss_s','method_disagreement_s']].head(20).to_string(index=False))
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(7, 5))
matched = primary.dropna(subset=['transition_loss_s','matched_control_loss_s'])
ax.scatter(matched.transition_loss_s, matched.matched_control_loss_s, s=10, alpha=.4)
ax.set(xlabel='Cross-fitted two-lap loss (s)', ylabel='Matched-control estimate (s)', title='Measurement agreement, not physical ground truth')
plt.show()
```
