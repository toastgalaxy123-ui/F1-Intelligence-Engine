# Stage 4: chronological model diagnostics and acceptance
Run the Stage 4 CLI first. These notebooks only inspect saved artifacts.


```python
from pathlib import Path
import json
import pandas as pd
root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent
run = root / 'artifacts/runs/stage4_pit_loss_v1'
if not (run / 'metrics.json').exists():
    raise FileNotFoundError('Run f1ri stage4 from the repository root first')
```

```python
metrics = json.loads((run / 'metrics.json').read_text())
rows = [{'model': model, 'split': split, **block['net_loss_proxy']} for model, splits in metrics['models'].items() for split, block in splits.items()]
print(pd.DataFrame(rows).to_string(index=False))
print(json.dumps(metrics['comparisons'], indent=2))
```

```python
print(json.dumps(json.loads((run / 'acceptance.json').read_text()), indent=2))
print('Software tests do not establish superiority. Do not optimize strategies using unsupported scenarios.')
```

```python
import matplotlib.pyplot as plt
image = plt.imread(run / 'plots/stage4_diagnostics.png')
fig, ax = plt.subplots(figsize=(20, 13))
ax.imshow(image)
ax.axis('off')
plt.show()
```
