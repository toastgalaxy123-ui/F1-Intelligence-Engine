# Stage 3.10–3.11 — Open-Loop Stint Trajectory Diagnostics (`D0`–`D10`)
Examines open-loop multi-lap and cumulative stint-time evaluation across horizons $H \in \{1, 3, 5, 10, 15, 20\}$, cold-start separation (`0A` vs `0B`), paired race bootstrap comparisons, and 2026 regulation-era transfer.

```python
import json
from pathlib import Path
import pandas as pd

metrics = json.loads(Path('../artifacts/runs/stage3_degradation_v1/metrics.json').read_text(encoding='utf-8'))
rows = []
for exp_id, m in metrics['experiments'].items():
    t = m['test_all']
    rows.append({
        'experiment': exp_id,
        'val_h5_cum_mae_s': m['val']['dataset_b_cumulative'].get('h5', {}).get('cum_race_weighted_mae_s'),
        'test_h1_lap_mae_s': t['dataset_a_lap'].get('h1', {}).get('race_weighted_mae_s'),
        'test_h5_cum_mae_s': t['dataset_b_cumulative'].get('h5', {}).get('cum_race_weighted_mae_s'),
        'test_h10_cum_mae_s': t['dataset_b_cumulative'].get('h10', {}).get('cum_race_weighted_mae_s'),
        'test_h15_cum_mae_s': t['dataset_b_cumulative'].get('h15', {}).get('cum_race_weighted_mae_s'),
        'test_h5_cov90': t['dataset_b_cumulative'].get('h5', {}).get('cum_coverage_90'),
    })
display(pd.DataFrame(rows))
```
