# Stage 4: pit data audit
Run the Stage 4 CLI first. These notebooks only inspect saved artifacts.


```python
from pathlib import Path
import json
import pandas as pd
root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent
run = root / 'artifacts/runs/stage4_pit_loss_v1'
if not (run / 'metrics.json').exists():
    raise FileNotFoundError('Run f1ri stage4 from the repository root first')
```

```python
events = pd.read_parquet(run / 'pit_events.parquet')
labels = pd.read_parquet(run / 'pit_event_labels.parquet')
print(events.groupby(['circuit_short_name', 'quality_category']).agg(events=('pit_event_id','nunique'), races=('session_key','nunique'), stationary_measured=('stop_duration_s','count')).to_string())
print(json.loads((run / 'data_audit.json').read_text()))
```

```python
print('Stationary missingness by season')
print(events.groupby('season').stop_duration_s.agg(['count','size','median']).to_string())
print('Measurement methods are uncertain counterfactual labels, not direct timing')
print(labels.groupby('measurement_method').transition_loss_s.agg(['count','median','std']).to_string())
```
